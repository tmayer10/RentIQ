```python
from pathlib import Path
import json
import random

random.seed(42)  # remove or change for different randomness

input_path = Path("test_apartments.json")
if not input_path.exists():
    raise FileNotFoundError(f"{input_path} does not exist")

with input_path.open("r", encoding="utf-8") as f:
    data = json.load(f)

# detect list of apartments in the loaded JSON
if isinstance(data, list):
    apartments = data
elif isinstance(data, dict):
    # prefer common keys, else pick the first list value found
    for key in ("apartments", "items", "data"):
        if key in data and isinstance(data[key], list):
            apartments = data[key]
            break
    else:
        list_vals = [v for v in data.values() if isinstance(v, list)]
        if not list_vals:
            raise ValueError("No list of apartments found in JSON")
        apartments = list_vals[0]
else:
    raise ValueError("Unexpected JSON structure")

n = 300
if len(apartments) >= n:
    sampled_apartments = random.sample(apartments, n)
else:
    sampled_apartments = apartments[:]  # return all if fewer than requested
    print(f"Only {len(apartments)} apartments available; returning all of them.")

# write result to a new file and keep variable for further notebook use
output_path = Path("sampled_300_apartments.json")
with output_path.open("w", encoding="utf-8") as f:
    json.dump(sampled_apartments, f, ensure_ascii=False, indent=2)

len(sampled_apartments)
```
